```python
# LOAD ALL THE INFORMATION THAT THE APP WILL REQUIRE
load_dotenv()
username = os.getenv("DB_USER")
password = os.getenv("DB_PW")
host = os.getenv("DB_HOST")
port = 3306
```

```python
#DB CONNECTION
db = pymysql.connect(host = host,
                     user = username,
                     password = password,
                     cursorclass = pymysql.cursors.DictCursor
)

# CURSOR OBJECT WILL EXECUTE ALL THE QUERIES AND FETCH THE DATA
cursor = db.cursor()
```

```python
#CREATE THE FIRST VERSION OF THE DEV DB
create_db = '''CREATE DATABASE kpi_database'''
cursor.execute(create_db)
```

### TABLE CREATION
1. Table **session** for getting id, id_user, date.  
2. Table **profile** this will pick up the required fields for the prompt, also it will give us researchable insights from the user.  
3. Table **project** this will register the project from what the user wants the KPIs.  
4. Table **question** this will register the question(prompt).  
5. Table **answers** this will pick the answers from the chat.  
6. Table **user** this will be more useful in the future. 


```python
create_table = '''
CREATE TABLE session (
    id INT NOT NULL auto_increment,
    id_user INT NOT NULL,
    date TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    primary key (id)
)
'''
cursor.execute(create_table)
db.commit()
```

```python
create_table = '''
CREATE TABLE profile (
    id_profile INT NOT NULL auto_increment,
    id_user INT NOT NULL,
    position VARCHAR(120),
    department VARCHAR(120),
    sector VARCHAR(120),
    primary key (id_profile)
)
'''
cursor.execute(create_table)
db.commit()
```

```python
create_table = '''
CREATE TABLE user (
    id_user INT NOT NULL,
    primary key (id_user)
)
'''
cursor.execute(create_table)
db.commit()
```

```python
create_table = '''
CREATE TABLE project (
    id_user INT NOT NULL,
    id_project INT NOT NULL auto_increment,
    project TEXT,
    primary key (id_project)
)
'''
cursor.execute(create_table)
db.commit()
```

```python
create_table = '''
CREATE TABLE question (
    id_question INT NOT NULL auto_increment,
    id INT NOT NULL,
    question TEXT NOT NULL,
    question_time TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    primary key (id_question)
)
'''
cursor.execute(create_table)
db.commit()
```

```python
create_table = '''
CREATE TABLE answer (
    id_answer INT NOT NULL auto_increment,
    id_question INT NOT NULL,
    id INT NOT NULL,
    answer TEXT,
    answer_time TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    primary key (id_answer)
)
'''
cursor.execute(create_table)
db.commit()
```
