## DB Actions and Review
With this notebook we'll be performing the actions that the APP requires, also we'll be reviewing that all the actions related with the DB are being executed properly.

```python
from dotenv import load_dotenv
import pymysql
import sqlalchemy
import os
from dotenv import load_dotenv
import pandas as pd
```

```python
def delete_session():
    try:

        delete_data = f"DELETE FROM session"
        cursor.execute(delete_data)
        
     
        reset_auto_increment = "ALTER TABLE session AUTO_INCREMENT = 1"
        cursor.execute(reset_auto_increment)
        

        db.commit()
        print("All sessions deleted and AUTO_INCREMENT reset to 1.")
    
    except Exception as e:
        db.rollback()  
        print(f"Error: {e}")
        
    finally:
        cursor.close() 
        db.close()  
```

```python
# LOAD ALL THE INFORMATION THAT THE APP WILL REQUIRE
load_dotenv()
username = os.getenv("DB_USER")
password = os.getenv("DB_PW")
host = os.getenv("DB_HOST")
port = 3306
```

```python
#DB CONNECTION
db = pymysql.connect(host = host,
                     user = username,
                     password = password,
                     cursorclass = pymysql.cursors.DictCursor
)

# CURSOR OBJECT WILL EXECUTE ALL THE QUERIES AND FETCH THE DATA
cursor = db.cursor()
```

```python
#CREATE THE FIRST VERSION OF THE DEV DB
create_db = '''CREATE DATABASE kpi_database'''
cursor.execute(create_db)
```

```python
#LET'S CHECK IS BEEN CREATED PROPERLY
cursor.execute('SHOW DATABASES')
cursor.fetchall()
```

```python
#WE'LL CREATE THE TABLE THAT IS GOING TO FETCH ID SESSION, PROMPT(QUESTION) AND ANSWER.
cursor.connection.commit()
use_db = ''' USE kpi_database'''
cursor.execute(use_db)

```

```python
create_table = '''
CREATE TABLE session (
    id INT NOT NULL auto_increment,
    prompt TEXT,
    answer TEXT,
    date TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    primary key (id)
)
'''
cursor.execute(create_table)
```

```python
prompt = "What is the best French cheese?"
answer = ''' 
The \"best\" French cheese can vary greatly depending on personal taste preferences. However, some of the most famous and highly 
regarded French cheeses include Brie de Meaux, Camembert de Normandie, Roquefort, Comté, 
and Époisses. Each of these cheeses has a unique flavor profile and texture. It's recommended to try a variety to find your personal favorite.
'''
insert_data = '''
INSERT INTO session (prompt, answer) 
VALUES (%s, %s)
'''
cursor.execute(insert_data, (prompt, answer))
```
